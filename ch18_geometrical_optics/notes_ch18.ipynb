# __CH-18 GEOMETRICAL OPTICS__

- To find path of light we deal with _geometrical optics_

## __18.1 REFLECTION AT SMOOTH SURFACES__

- When a ray of light approaches a smooth polished surface and the light ray bounces back, it is called the reflection of light

__Laws of Reflection__

1. The angle of incidence is always equal to the angle of reflection

2. The incident ray, reflected ray and normal to the reflected surface at the point of incidence, all lie in the same plane

## __18.2 SPHERICAL MIRRORS__

- Spherical mirrors are made of glass

- One side is silvered

- Reflection takes place on the other side

- If reflection takes place on the convex surface, it is a _convex mirror_

- If reflection takes place on the concave surface, it is a _concave mirror_


- The centre of the sphere of which the mirror forms a part is called _centre of curvature_

- The radius of this sphere is called _radius of curvature_

- The point on the mirror  at the middle of the surface is called _pole_

- The line joining the pole and the centre of curvature is called _principal axis_

__Focus__

- The point where the reflected rays converge is called the focus of the mirror

- The plane through the focus and perpendicular to the principal axis is called the _focal plane_

- The distance of the focus from the pole is called _focal length_

__Paraxial Rays__

- A ray close to the principal axis is called a paraxial ray

__Image Tracing__

- If the reflected rays intersect, the point of intersection is the _real image_

- If the rays diverge after reflection, a virtual image is formed
